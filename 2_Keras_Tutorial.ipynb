# Keras深度學習初體驗1
利用keras套件，產生一個32個神經元的深度學習模型進行訓練


```
from tensorflow import keras
from keras.models import Sequential
from keras.layers import Dense

import numpy as np
data = np.random.random((500, 10)) #隨機產生500筆資料，每筆資料包含十個float數字
labels = np.random.randint(2, size=(500, 1)) #隨機產生500筆資料，每筆資料上限為2(意思就是0 或 1)

print(np.shape(data))
print(data[0])
print(np.shape(labels))
print(labels[0])

model = Sequential()
model.add(Dense(128, activation='relu', input_dim=10))
model.add(Dense(1, activation='sigmoid'))
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])#設定訓練方式，使用Adam優化器

model.fit(data, labels, epochs=35, batch_size=16)

score=model.evaluate(data, labels)
print(score[1])#印出準確率（0~1的小數）
```
